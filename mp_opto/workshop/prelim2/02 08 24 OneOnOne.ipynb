# bulk population structure, during time of laser (onset->offset)

```python
output = pload('../picklejar/bulkQuantify25post.pickle')
avg_tuple = []
fractional = []
l_divergence = []
c_divergence = []
inactivate = []
for idx, session in enumerate(output):
    if session == 0:
        print(f'{idx} not working')
        continue
    #avg_tuple.append(session[0])
    #fractional.append(session[1])
    l_divergence.append(session[2])
    c_divergence.append(session[3])
    inactivate.append(session[4])
    
downstream_laser = []
downstream_ctrl = []
upstream_laser = []
upstream_ctrl = []

downstream_cfa=[]
downstream_rfa=[]
upstream_cfa=[]
upstream_rfa=[]
for idx, region in enumerate(inactivate):
    if region == 'rfa':
        downstream_laser.append(l_divergence[idx][0])
        downstream_ctrl.append(c_divergence[idx][0])
        upstream_laser.append(l_divergence[idx][1])
        upstream_ctrl.append(c_divergence[idx][1])
        
        downstream_cfa.append(l_divergence[idx][0])
        upstream_rfa.append(l_divergence[idx][1])
        
    else:
        downstream_laser.append(l_divergence[idx][1])
        downstream_ctrl.append(c_divergence[idx][1])
        upstream_laser.append(l_divergence[idx][0])
        upstream_ctrl.append(c_divergence[idx][0])
        
        downstream_rfa.append(l_divergence[idx][1])
        upstream_cfa.append(l_divergence[idx][0])
```

```python
output = pload('../picklejar/bulkSweepLags_post50.pickle')
```

```python
laser_lag_dict = {}
ctrl_lag_dict = {}

laser_cfa_dict = {}
laser_rfa_dict = {}
ctrl_cfa_dict = {}
ctrl_rfa_dict = {}
nlags = np.arange(2,22,2)
for lag in nlags:
    laser_lag_dict[lag] = []
    ctrl_lag_dict[lag] = []
    laser_cfa_dict[lag] = []
    ctrl_cfa_dict[lag] = []
    laser_rfa_dict[lag] = []
    ctrl_rfa_dict[lag] = []

for idx, session in enumerate(output):
    if session == 0:
        continue
    for model in session:
        laser_lag_dict[model[3]].append(model[1])
        ctrl_lag_dict[model[3]].append(model[2])
        if model[4] == 'cfa':
            laser_rfa_dict[model[3]].append(model[1])
            ctrl_rfa_dict[model[3]].append(model[2])
        else:
            laser_cfa_dict[model[3]].append(model[1])
            ctrl_cfa_dict[model[3]].append(model[2])
            
        
```

```python
laser_lags = []
laser_lags_std = []
ctrl_lags = []
ctrl_lags_std = []

laser_cfa = []
laser_cfa_std = []
ctrl_cfa = []
ctrl_cfa_std = []

laser_rfa = []
laser_rfa_std = []
ctrl_rfa = []
ctrl_rfa_std = []
for key in laser_lag_dict.keys():
    laser_lags.append(np.average(laser_lag_dict[key]))
    laser_lags_std.append(np.std(laser_lag_dict[key]))
    ctrl_lags.append(np.average(ctrl_lag_dict[key]))
    ctrl_lags_std.append(np.std(ctrl_lag_dict[key]))
    
    laser_cfa.append(np.average(laser_cfa_dict[key]))
    laser_cfa_std.append(np.std(laser_cfa_dict[key]))
    ctrl_cfa.append(np.average(ctrl_cfa_dict[key]))
    ctrl_cfa_std.append(np.std(ctrl_cfa_dict[key]))
    
    laser_rfa.append(np.average(laser_rfa_dict[key]))
    laser_rfa_std.append(np.std(laser_rfa_dict[key]))
    ctrl_rfa.append(np.average(ctrl_rfa_dict[key]))
    ctrl_rfa_std.append(np.std(ctrl_rfa_dict[key]))
    
```
